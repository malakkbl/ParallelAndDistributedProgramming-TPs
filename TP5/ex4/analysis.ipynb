# TP5 - Exercise 4: MPI Matrix-Vector Product Performance Analysis

This notebook analyzes the performance of the MPI-based parallel matrix-vector multiplication implementation.

## Objective
Compare sequential and parallel execution times, and evaluate speedup and efficiency for different matrix sizes and process counts.

## 1. Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Set style for better-looking plots
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11
```

## 2. Load Performance Data

```python
# Load the CSV file
df = pd.read_csv('results.csv')

# Remove duplicate entries (keep last occurrence for each matrix_size + num_processes)
df = df.drop_duplicates(subset=['Matrix_Size', 'Num_Processes'], keep='last').reset_index(drop=True)

# Display the first few rows
print("Performance Data:")
print(df.head(10))
print(f"\nTotal rows: {len(df)}")
print(f"Matrix sizes: {sorted(df['Matrix_Size'].unique())}")
print(f"Process counts: {sorted(df['Num_Processes'].unique())}")
# Debug: check row counts per matrix size
print(f"\nRows per matrix size:")
for size in sorted(df['Matrix_Size'].unique()):
    count = len(df[df['Matrix_Size'] == size])
    print(f"  Size {size}: {count} rows")
```

## 3. Execution Time Analysis

```python
# Plot execution times for each matrix size
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Execution Time Analysis: Sequential vs Parallel', fontsize=16, fontweight='bold')

matrix_sizes = sorted(df['Matrix_Size'].unique())

for idx, size in enumerate(matrix_sizes):
    ax = axes[idx // 3, idx % 3]
    
    # Filter data for this matrix size
    data = df[df['Matrix_Size'] == size].copy()
    data = data.sort_values('Num_Processes')
    
    processes = data['Num_Processes'].values
    seq_time = data['Sequential_Time'].values
    par_time = data['Parallel_Time'].values
    
    # Plot
    ax.plot(processes, seq_time, 'o-', label='Sequential', linewidth=2, markersize=8, color='blue')
    ax.plot(processes, par_time, 's-', label='Parallel', linewidth=2, markersize=8, color='red')
    
    ax.set_xlabel('Number of Processes', fontweight='bold')
    ax.set_ylabel('Execution Time (seconds)', fontweight='bold')
    ax.set_title(f'Matrix Size: {size}x{size}', fontweight='bold')
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_xticks(processes)

# Remove extra subplot
fig.delaxes(axes[1, 2])

plt.tight_layout()
plt.savefig('execution_time_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Execution time comparison plot saved as 'execution_time_comparison.png'")
```

## 4. Speedup Analysis

```python
# Plot speedup for all matrix sizes
plt.figure(figsize=(12, 7))

colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']
markers = ['o', 's', '^', 'D', 'v']

for idx, size in enumerate(matrix_sizes):
    data = df[df['Matrix_Size'] == size].copy()
    data = data.sort_values('Num_Processes')
    
    processes = data['Num_Processes'].values
    speedup = data['Speedup'].values
    
    plt.plot(processes, speedup, marker=markers[idx], label=f'Size {size}x{size}', 
             linewidth=2.5, markersize=10, color=colors[idx])

# Plot ideal speedup line
ideal_processes = np.array([1, 2, 4, 8, 16])
plt.plot(ideal_processes, ideal_processes, 'k--', label='Ideal Speedup', linewidth=2, alpha=0.7)

plt.xlabel('Number of Processes', fontsize=13, fontweight='bold')
plt.ylabel('Speedup', fontsize=13, fontweight='bold')
plt.title('Speedup vs Number of Processes', fontsize=15, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.xticks([1, 2, 4, 8, 16])
plt.tight_layout()
plt.savefig('speedup_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Speedup comparison plot saved as 'speedup_comparison.png'")
```

## 5. Efficiency Analysis

```python
# Plot efficiency for all matrix sizes
plt.figure(figsize=(12, 7))

for idx, size in enumerate(matrix_sizes):
    data = df[df['Matrix_Size'] == size].copy()
    data = data.sort_values('Num_Processes')
    
    processes = data['Num_Processes'].values
    efficiency = data['Efficiency'].values
    
    plt.plot(processes, efficiency, marker=markers[idx], label=f'Size {size}x{size}', 
             linewidth=2.5, markersize=10, color=colors[idx])

# Plot 100% efficiency line
plt.axhline(y=100, color='k', linestyle='--', linewidth=2, alpha=0.7, label='Ideal Efficiency (100%)')

plt.xlabel('Number of Processes', fontsize=13, fontweight='bold')
plt.ylabel('Efficiency (%)', fontsize=13, fontweight='bold')
plt.title('Parallel Efficiency vs Number of Processes', fontsize=15, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.xticks([1, 2, 4, 8, 16])
plt.ylim(0, 110)
plt.tight_layout()
plt.savefig('efficiency_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Efficiency comparison plot saved as 'efficiency_comparison.png'")
```

## 6. Parallel Time Analysis

```python
# Plot parallel execution time for all matrix sizes
plt.figure(figsize=(12, 7))

for idx, size in enumerate(matrix_sizes):
    data = df[df['Matrix_Size'] == size].copy()
    data = data.sort_values('Num_Processes')
    
    processes = data['Num_Processes'].values
    par_time = data['Parallel_Time'].values
    
    plt.plot(processes, par_time, marker=markers[idx], label=f'Size {size}x{size}', 
             linewidth=2.5, markersize=10, color=colors[idx])

plt.xlabel('Number of Processes', fontsize=13, fontweight='bold')
plt.ylabel('Parallel Execution Time (seconds)', fontsize=13, fontweight='bold')
plt.title('Parallel Execution Time vs Number of Processes', fontsize=15, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.xticks([1, 2, 4, 8, 16])
plt.tight_layout()
plt.savefig('parallel_time_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Parallel time comparison plot saved as 'parallel_time_comparison.png'")
```

## 7. Combined Analysis Dashboard

```python
# Create a comprehensive dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('MPI Matrix-Vector Product: Comprehensive Performance Analysis', 
             fontsize=16, fontweight='bold', y=0.995)

# subplot 1: Parallel execution time
ax1 = axes[0, 0]
for idx, size in enumerate(matrix_sizes):
    data = df[df['Matrix_Size'] == size].sort_values('Num_Processes')
    ax1.plot(data['Num_Processes'], data['Parallel_Time'], marker=markers[idx], 
             label=f'{size}x{size}', linewidth=2.5, markersize=8, color=colors[idx])
ax1.set_xlabel('Number of Processes', fontweight='bold')
ax1.set_ylabel('Parallel Time (seconds)', fontweight='bold')
ax1.set_title('Parallel Execution Time', fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)
ax1.set_xticks([1, 2, 4, 8, 16])

# subplot 2: Speedup
ax2 = axes[0, 1]
for idx, size in enumerate(matrix_sizes):
    data = df[df['Matrix_Size'] == size].sort_values('Num_Processes')
    ax2.plot(data['Num_Processes'], data['Speedup'], marker=markers[idx], 
             label=f'{size}x{size}', linewidth=2.5, markersize=8, color=colors[idx])
ideal_p = np.array([1, 2, 4, 8, 16])
ax2.plot(ideal_p, ideal_p, 'k--', label='Ideal', linewidth=2, alpha=0.7)
ax2.set_xlabel('Number of Processes', fontweight='bold')
ax2.set_ylabel('Speedup', fontweight='bold')
ax2.set_title('Speedup', fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)
ax2.set_xticks([1, 2, 4, 8, 16])

# subplot 3: Efficiency
ax3 = axes[1, 0]
for idx, size in enumerate(matrix_sizes):
    data = df[df['Matrix_Size'] == size].sort_values('Num_Processes')
    ax3.plot(data['Num_Processes'], data['Efficiency'], marker=markers[idx], 
             label=f'{size}x{size}', linewidth=2.5, markersize=8, color=colors[idx])
ax3.axhline(y=100, color='k', linestyle='--', linewidth=2, alpha=0.7, label='Ideal (100%)')
ax3.set_xlabel('Number of Processes', fontweight='bold')
ax3.set_ylabel('Efficiency (%)', fontweight='bold')
ax3.set_title('Parallel Efficiency', fontweight='bold')
ax3.legend()
ax3.grid(True, alpha=0.3)
ax3.set_xticks([1, 2, 4, 8, 16])
ax3.set_ylim(0, 110)

# subplot 4: Execution time comparison (Sequential vs Best Parallel)
ax4 = axes[1, 1]
sizes_list = []
seq_times = []
best_par_times = []
for size in matrix_sizes:
    data = df[df['Matrix_Size'] == size].sort_values('Num_Processes')
    sizes_list.append(size)
    seq_times.append(data['Sequential_Time'].iloc[0])
    best_par_times.append(data['Parallel_Time'].min())

x_pos = np.arange(len(sizes_list))
width = 0.35
ax4.bar(x_pos - width/2, seq_times, width, label='Sequential', color='blue', alpha=0.7)
ax4.bar(x_pos + width/2, best_par_times, width, label='Best Parallel', color='red', alpha=0.7)
ax4.set_xlabel('Matrix Size', fontweight='bold')
ax4.set_ylabel('Execution Time (seconds)', fontweight='bold')
ax4.set_title('Sequential vs Best Parallel Time', fontweight='bold')
ax4.set_xticks(x_pos)
ax4.set_xticklabels([f'{s}x{s}' for s in sizes_list], rotation=45)
ax4.legend()
ax4.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('combined_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Combined analysis dashboard saved as 'combined_analysis.png'")
```

## 8. Statistical Summary

```python
# Summary statistics
print("="*80)
print("PERFORMANCE SUMMARY")
print("="*80)

for size in matrix_sizes:
    data = df[df['Matrix_Size'] == size].copy()
    data = data.sort_values('Num_Processes')
    
    print(f"\nMatrix Size: {size}x{size}")
    print("-" * 80)
    print(f"{'Processes':<12} {'Seq Time (s)':<15} {'Par Time (s)':<15} {'Speedup':<12} {'Efficiency (%)':<15}")
    print("-" * 80)
    
    for _, row in data.iterrows():
        print(f"{row['Num_Processes']:<12} {row['Sequential_Time']:<15.6f} {row['Parallel_Time']:<15.6f} "
              f"{row['Speedup']:<12.3f} {row['Efficiency']:<15.2f}")
    
    # Best performance
    best_speedup_row = data.loc[data['Speedup'].idxmax()]
    best_efficiency_row = data.loc[data['Efficiency'].idxmax()]
    
    print(f"\n  → Best Speedup: {best_speedup_row['Speedup']:.3f}x with {int(best_speedup_row['Num_Processes'])} processes")
    print(f"  → Best Efficiency: {best_efficiency_row['Efficiency']:.2f}% with {int(best_efficiency_row['Num_Processes'])} processes")

print("\n" + "="*80)
```

## 9. Key Observations

### Expected Behavior:
1. **Speedup**: Should increase with more processes, but may be sublinear due to communication overhead
2. **Efficiency**: Typically decreases as the number of processes increases (communication overhead grows)
3. **Matrix Size Impact**: Larger matrices should show better parallel efficiency due to higher computation-to-communication ratio
4. **Oversubscription Effect**: Performance with 8 and 16 processes may be affected since they exceed physical core count

### Analysis Points:
- Compare actual speedup vs ideal linear speedup
- Identify the optimal number of processes for each matrix size
- Evaluate the impact of MPI communication overhead (MPI_Scatterv, MPI_Gatherv, MPI_Bcast)
- Discuss the effect of load balancing for non-divisible matrix sizes
